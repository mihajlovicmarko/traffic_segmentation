```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
def VarianceFromData(data_array : np.array):
    return np.sum(np.square(np.mean(data_array) - data_array)) / data_array.shape[0]
```

```python
n_trials = 20000
array_of_means = np.zeros(n_trials)
array_of_variences = np.zeros(n_trials)
for i in range(n_trials):
    samples = np.sort(np.random.normal(0, 10, size = 1000))
    array_of_means[i] = np.mean(samples)
    array_of_variences[i] = VarianceFromData(samples)
    
    
#plt.hist(samples, bins=20, density=True, alpha=0.6, color='g')
plt.hist(array_of_means, bins=15, density=True, alpha=0.6, color='g')
```

```python
Sn = VarianceFromData(samples)
t_dist = np.sort(np.random.normal(np.mean(samples), np.sqrt(Sn / samples.shape[0]), size = 100))
plt.hist(t_dist, bins=20, density=True, alpha=0.6, color='g')
```
